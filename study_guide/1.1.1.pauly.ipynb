## Pauli
https://quantum.cloud.ibm.com/docs/en/api/qiskit/qiskit.quantum_info.Pauli

```python
from qiskit.quantum_info import Pauli

P = Pauli('-iXYZ')

print('P[0] =', repr(P[0]))
print('P[1] =', repr(P[1]))
print('P[2] =', repr(P[2]))
print('P[:] =', repr(P[:]))
print('P[::-1] =', repr(P[::-1]))
```

```python
p = Pauli('IZ')
print(p.to_matrix())
```

```python
from qiskit.quantum_info import Pauli

for s in ['II', 'IZ', 'ZI', 'ZZ', 'XX', '-II', '-ZI']:
    p = Pauli(s)
    print(f"\nPauli: {s}")
    print(p.to_matrix())
```
