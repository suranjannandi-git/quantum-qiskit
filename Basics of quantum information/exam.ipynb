```python
import numpy as np
from qiskit.quantum_info import Operator

# Pauli-Z matrix
Z = Operator([[1, 0],
              [0, -1]])

M = Z.data

# Hermitian: M† = M
is_hermitian = np.allclose(M.conj().T, M)

# Unitary: M†M = I
is_unitary = np.allclose(M.conj().T @ M, np.eye(M.shape[0]))

print("Hermitian:", is_hermitian)
print("Unitary:", is_unitary)
```

```python
import numpy as np

def matrix_properties(M):
    M = np.asarray(M)
    I = np.eye(M.shape[0], dtype=complex)

    return {
        "Hermitian": np.allclose(M.conj().T, M),
        "Unitary": np.allclose(M.conj().T @ M, I),
    }

# Example
from qiskit.quantum_info import Operator

# This approach works for X, Y, Z, H, S, T, CNOT, or any custom matrix or Qiskit Operator.
props = matrix_properties(Operator.from_label("S").data)
print(props)
```

```python
# Q5
import numpy as np
from qiskit.quantum_info import Statevector

psi = Statevector([
    np.sqrt(2)/3,
    -1/3,
    np.sqrt(2)/3,
    2/3
])

psi0 = np.array([
    1/np.sqrt(3),
    np.sqrt(2/3)
], dtype=complex)

psi1 = np.array([
    np.sqrt(2/3),
    -1/np.sqrt(3)
], dtype=complex)

print(np.vdot(psi0, psi0))
print(np.vdot(psi1, psi1))
print(np.vdot(psi0, psi1))

P0 = np.outer(psi0, psi0.conj())

PI0 = np.kron(np.eye(2), P0)

prob = np.vdot(psi.data, PI0 @ psi.data).real
print(prob)
```

```python
# Q6
import numpy as np
from qiskit.quantum_info import Statevector

psi0 = Statevector([5/13, 12/13])
psi1 = Statevector([12/13, -5/13])

inner_product = np.vdot(psi0.data, psi1.data)

# Convert to SymPy matrices for display
psi0_sym = sp.Matrix(psi0.data).applyfunc(sp.nsimplify)
psi1_sym = sp.Matrix(psi1.data).applyfunc(sp.nsimplify)
inner_sym = sp.nsimplify(inner_product)

display(Math(
    r"\langle \psi_0 | \psi_1 \rangle = "
    + sp.latex(psi0_sym.T)
    + sp.latex(psi1_sym)
    + " = "
    + sp.latex(inner_sym)
))

print("Inner product =", inner_product)
print("Orthogonal?", np.isclose(inner_product, 0))
```
