#### 1. Import required libraries

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
```

#### 2. Create the Bell states

```python
# A. Bell state Φ⁺

qc = QuantumCircuit(2)

qc.h(0)
qc.cx(0, 1)

qc.measure_all()
qc.draw("mpl")
```

```python
# B. Bell state Φ⁻

qc = QuantumCircuit(2)

qc.h(0)
qc.z(0)
qc.cx(0, 1)

qc.measure_all()
qc.draw("mpl")
```

```python
# C. Bell state Ψ⁺

qc = QuantumCircuit(2)

qc.h(0)
qc.x(1)
qc.cx(0, 1)

qc.measure_all()
qc.draw("mpl")
```

```python
# D. Bell state Ψ⁻

qc = QuantumCircuit(2)

qc.h(0)
qc.x(1)
qc.z(0)
qc.cx(0, 1)

qc.measure_all()

qc.draw("mpl")
```

#### # 3. Execute and observe measurement results

```python
sampler = StatevectorSampler()

result = sampler.run([qc], shots=1024).result()

counts = result[0].data.meas.get_counts()

print(counts)
```

#### 4. Create a GHZ state (3 qubits)

```python
qc = QuantumCircuit(3)

qc.h(0)
qc.cx(0, 1)
qc.cx(0, 2)

qc.measure_all()

qc.draw("mpl")
```

```python
# Execute 
result = sampler.run([qc], shots=1024).result()

counts = result[0].data.meas.get_counts()

print(counts)
```

#### 5. Create a W state (3 qubits)

```python
qc = QuantumCircuit(3)

qc.ry(2 * 0.9553166, 0)
qc.ch(0, 1)
qc.cx(1, 2)
qc.cx(0, 1)

qc.measure_all()

qc.draw("mpl")
```

```python
result = sampler.run([qc], shots=1024).result()

counts = result[0].data.meas.get_counts()

print(counts)
```
