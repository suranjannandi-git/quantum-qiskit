```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler # Updated import
import numpy as np

# Create a Quantum Circuit with 4 qubits and 4 classical bits
qc = QuantumCircuit(4, 4)

# Set input qubits to represent |1> and |1>
qc.x(0)  # Flips qubit 0 to 1
qc.x(1)  # Flips qubit 1 to 1

# --- Half-Adder Logic Gates ---
qc.ccx(0, 1, 3)  # CCNOT: calculates the Carry bit and stores in qubit 3
qc.cx(0, 1)      # CNOT: calculates the Sum bit and stores in qubit 1
qc.cx(1, 2)      # CNOT: moves the Sum to qubit 2

# Measure the outputs
qc.measure(1, 2) # Measure the Sum qubit into classical bit 2
qc.measure(3, 3) # Measure the Carry qubit into classical bit 3

# Instantiate the modern V2 sampler
sampler = StatevectorSampler()

# Run the job using a PUB (Primitive Unified Block) tuple: (circuit,)
job = sampler.run([(qc,)])
result = job.result()

# Extract bitstring counts from the first PUB result
pub_result = result[0]
counts = pub_result.data.c.get_counts()

# Print the final dictionary of outcomes
print("Measured results:", counts)
```
