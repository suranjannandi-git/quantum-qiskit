```python
from qiskit import *
from qiskit.visualization import plot_bloch_multivector, visualize_transition, plot_histogram
from qiskit_aer import Aer
from qiskit.quantum_info import Statevector
```

```python
# initialize a 2 qubit quantum circuit
qc = QuantumCircuit(2)
# by default the qubits in the circuit will be in |00> state 
# represented as [1, 0, 0, 0] in the state vector
```

```python
# applying hadamard gate on all qubits
qc.h(0)
qc.x(1)
```

```python
qc.draw('mpl')  # using matplotlib library, draw the quantum circuit
```

```python
# %pip install git+https://github.com/qiskit-community/qiskit-textbook.git#subdirectory=qiskit-textbook-src
```

```python
backend = Aer.get_backend("statevector_simulator")
state = backend.run(qc).result().get_statevector()

# Plot Bloch sphere
plot_bloch_multivector(state)
```

```python
from qiskit_textbook.tools import array_to_latex    
```

```python
from qiskit_textbook.tools import array_to_latex
array_to_latex(state, pretext="\\text{Statevector} = ")
```

```python
from IPython.display import display
from qiskit.visualization import array_to_latex

display(
    array_to_latex(
        state,
        prefix="\\text{UnitaryMatrix} = "
    )
)
```

```python
backend = Aer.get_backend("unitary_simulator")
state = backend.run(qc).result().get_unitary()

# from qiskit_textbook.tools import array_to_latex
from qiskit.visualization import array_to_latex
display(array_to_latex(state, prefix="\\text{UnitaryMatrix} = "))
```
