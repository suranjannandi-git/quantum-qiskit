# Deutsch-Jozsa Algorithm

```python
from qiskit import *
from qiskit_aer import AerSimulator
```

```python
qc = QuantumCircuit(3, 2)
```

```python
qc.x(2)  # Apply X gate to the ausiliary qubit (qubit 2)

qc.h(0)  # Apply H gate to qubit 0
qc.h(1)  # Apply H gate to qubit 1
qc.h(2)  # Apply H gate to qubit 2
qc.barrier()
```

```python
# simulate a constant function, do nothing as it will always give a constant output
# do nothing

# simulate a balanced function, f(x) = x
qc.cx(0, 2)  # Apply CNOT gate with qubit 0
qc.cx(1, 2)  # Apply CNOT gate with qubit 1 
qc.barrier()
```

```python
qc.h(0)  # Apply H gate to qubit 0
qc.h(1)  # Apply H gate to qubit 1
```

```python
qc.measure(0, 0)  # Measure qubit 0 and store the result in classical bit 0
qc.measure(1, 1)  # Measure qubit 1 and store the result in classical bit 1
```

```python
qc.draw()
```

```python
simulator = AerSimulator()
result = simulator.run(qc, backend=simulator).result()
print(result.get_counts())
```
