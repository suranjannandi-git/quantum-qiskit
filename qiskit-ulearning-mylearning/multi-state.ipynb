```python
from qiskit import *
from qiskit.visualization import plot_bloch_multivector, visualize_transition, plot_histogram
from qiskit_aer import Aer
from qiskit.quantum_info import Statevector
```

## Multi State-Gate

```python
# initialize a 3 qubit quantum circuit
qc = QuantumCircuit(3)
# by default the qubits in the circuit will be in |000> state 
# represented as [1, 0, 0, 0, 0, 0, 0, 0] in the state vector
```

```python
# get the current state of the qubit after applying the gate and plot it 
after_state = Statevector.from_instruction(qc)
plot_bloch_multivector(after_state)
```

```python
# applying hadamard gate on all qubits
qc.h([0, 1, 2])
```

```python
qc.draw('mpl')  # using matplotlib library, draw the quantum circuit
```

```python
backend = Aer.get_backend("statevector_simulator")
state = backend.run(qc).result().get_statevector()

# Plot Bloch sphere
plot_bloch_multivector(state)
```
