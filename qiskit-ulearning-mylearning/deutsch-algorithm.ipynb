# Deutsch's Algorithm 

```python
from qiskit import *
from qiskit_aer import AerSimulator
```

```python
qc = QuantumCircuit(2, 1)       
```

```python
qc.x(1)  # Apply X gate to qubit 1
qc.h(0)  # Apply H gate to qubit 0
qc.h(1)  # Apply H gate to qubit 1
qc.barrier()  # Add a barrier to separate the oracle from the rest of the circuit
```

```python
# Create a constant oracle for the function f(x) = 0 
# do nothing as it will always give a constant output 

# Create a balanced oracle for the function f(x) = x
qc.cx(0, 1)  # Apply CNOT gate with qubit 0 as control and qubit 1 as target

qc.barrier()  # Add a barrier to separate the oracle from the rest of the circuit   
```

```python
qc.h(0)  # Apply H gate to qubit 0
qc.barrier()  # Add a barrier to separate the oracle from the rest of the circuit
```

```python
qc.measure(0, 0)  # Measure qubit 0 and store the result in classical bit 0
```

```python
qc.draw()
```

```python
simulator = AerSimulator()
result = simulator.run(qc, backend=simulator).result()
print(result.get_counts())
```
