# Tiffoli Gate

```python
from qiskit import *
from qiskit.visualization import plot_bloch_multivector, visualize_transition, plot_histogram
from qiskit_aer import Aer
from qiskit.quantum_info import Statevector
from qiskit.quantum_info import Operator
from qiskit.visualization import array_to_latex
from IPython.display import display
```

```python
qc = QuantumCircuit(3)
c1 = 0
c2 = 1
t = 2
```

```python
qc.ccx(c1, c2, t)
qc.draw('mpl')
```

```python
backend = Aer.get_backend('unitary_simulator')
output = backend.run(qc).result().get_unitary()
```

```python
# Display the matrix in LaTeX
display(array_to_latex(output, prefix="\\text{Circuit output = }"))
```

```python
backend = Aer.get_backend('statevector_simulator')
output = backend.run(qc).result().get_statevector()
```

```python
array_to_latex(output, prefix="\\text{Statevector} = ")
```
