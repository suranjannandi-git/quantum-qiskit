# Phase Oracle

```python
from qiskit import *
```

```python
# three qubits, q0 = A, q1 = B, q2 = C, Auxiliary qubit 
qc = QuantumCircuit(3)
# have an example case A=1, B=0, C=|->
# apply X gate to make A = 1
qc.x(0)  # now A=|1>
# convert C to |-> state. At first convert C to |1> and then apply H gate to convert |1> to |-> state.
qc.x(2)  # now C=|1>    
qc.h(2)  # now C=|->
```

```python
# OR Oracle implementation using CNOT and Toffoli gates.
# Step 1: C XOR A (C CNOT A)
qc.cx(0, 2)  

# Step 2: C XOR B (C CNOT B)
qc.cx(1, 2)  

# Step 3: C XOR (A and B) = CCX (0, 1, 2) (C Toffoli A B)
qc.ccx(0, 1, 2)
print(qc.draw())
```

```python
from qiskit.quantum_info import Statevector
state = Statevector.from_instruction(qc)
print("Final state Dict:", state.to_dict())
```
